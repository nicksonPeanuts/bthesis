```python
# LV CLASSICO

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parameters
c=2




U0 = float(input("Enter a positive value for U(0): "))

V0 = float(input("Enter a positive value for V(0): "))

c = float(input("Enter a positive value for C: "))

Tmax = float(input("Enter a positive value for the max simulation time : "))


# CLASSICAL LOTKA VOLTERRA
def lotka_volterra(t, y):
    U, V = y
    dU_dt = U*(1  -   V)
    dV_dt = c*V*( -1 + U  )
    return [dU_dt, dV_dt]

# Initial conditions
#U0 = 0.1  # Initial prey population
#V0 = 0.1  # Initial predator population

y0 = [U0, V0]

# Time span for the simulation
t_span = (0, Tmax)  
t_eval = np.linspace(t_span[0], t_span[1], 10000)  

# Solve the ODEs using solve_ivp
solution = solve_ivp(
    lotka_volterra, t_span, y0, t_eval=t_eval, method='RK45',
    rtol=1e-6, atol=1e-9
)

# Extract solutions
U = solution.y[0]
V = solution.y[1]
t = solution.t

# Plotting the results
plt.figure(figsize=(14, 6))

# Plot U(t) and V(t) over time
plt.subplot(1, 2, 1)
plt.plot(t, U, label='Prey Population U(t)', color='blue')
plt.plot(t, V, label='Predator Population V(t)', color='red')
plt.xlabel('Time')
plt.ylabel('Population')
plt.title('Population Dynamics Over Time')
plt.legend()
plt.grid(True)

# Phase plane plot (V vs U)
plt.subplot(1, 2, 2)
plt.plot(U, V, color='green')
plt.xlabel('Prey Population U')
plt.ylabel('Predator Population V')
plt.title('Phase Plane Trajectory (V vs U)')
plt.grid(True)

plt.tight_layout()
plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parameters





c = float(input("Enter a positive value for c : "))

a = float(input("Enter a positive value for a (<1): "))



# System of differential equations
def lotka_volterra(t, y):
    U, V = y
    dU_dt = U*( (1-a*U)  -   V)
    dV_dt = c*V*( -1 + U   )
    return [dU_dt, dV_dt]

# Initial conditions
U0 = 0.1  # Initial prey population
V0 = 0.1  # Initial predator population
y0 = [U0, V0]

# Time span for the simulation
t_span = (0, 200)  
t_eval = np.linspace(t_span[0], t_span[1], 10000)  

# Solve the ODEs using solve_ivp
solution = solve_ivp(
    lotka_volterra, t_span, y0, t_eval=t_eval, method='RK45',
    rtol=1e-6, atol=1e-9
)

# Extract solutions
U = solution.y[0]
V = solution.y[1]
t = solution.t

# Plotting the results
plt.figure(figsize=(14, 6))

# Plot U(t) and V(t) over time
plt.subplot(1, 2, 1)
plt.plot(t, U, label='Prey Population U(t)', color='blue')
plt.plot(t, V, label='Predator Population V(t)', color='red')
plt.xlabel('Time')
plt.ylabel('Population')
plt.title('Population Dynamics Over Time')
plt.legend()
plt.grid(True)

# Phase plane plot (V vs U)
plt.subplot(1, 2, 2)
plt.plot(U, V, color='green')
plt.xlabel('Prey Population U')
plt.ylabel('Predator Population V')
plt.title('Phase Plane Trajectory (V vs U)')
plt.grid(True)

plt.tight_layout()
plt.show()

```
