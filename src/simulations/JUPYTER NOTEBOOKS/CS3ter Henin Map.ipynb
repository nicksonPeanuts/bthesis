```python
import numpy as np
import matplotlib.pyplot as plt

# Parametri classici della mappa di Hénon
a = 1.4
b = 0.3

def henon_step(state, a=a, b=b):
    """
    Un passo della mappa di Hénon.
    state: array-like [x, y]
    """
    x, y = state
    x_next = 1.0 - a * x**2 + y
    y_next = b * x
    return np.array([x_next, y_next])

# Parametri di simulazione
N_total = 100_000     # numero totale di iterazioni
N_transient = 1_000   # iterazioni da scartare come transiente

# Condizione iniziale
state = np.array([0.0, 0.0], dtype=float)

# Lista per salvare i punti dopo il transiente
trajectory = []

for n in range(N_total):
    state = henon_step(state)
    if n >= N_transient:
        trajectory.append(state.copy())

trajectory = np.array(trajectory)
x = trajectory[:, 0]
y = trajectory[:, 1]

# Plot dell'attrattore nel piano (x, y)
plt.figure(figsize=(6, 6))
plt.scatter(x, y, s=0.1, linewidths=0)

plt.title("Attrattore della mappa di Hénon (a=1.4, b=0.3)")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.axis("equal")
plt.tight_layout()
plt.show()
```
